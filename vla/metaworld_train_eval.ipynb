```python
# @title 3. Fine-tune SmolVLA sur Meta-World (commande officielle de la doc LeRobot)
!lerobot-train \
    --policy.type=smolvla \
    --policy.load_vlm_weights=true \
    --policy.push_to_hub=false \
    --dataset.repo_id=lerobot/metaworld_mt50 \
    --env.type=metaworld \
    --env.task=assembly-v3,dial-turn-v3,handle-press-side-v3 \
    --output_dir=./outputs/metaworld_smolvla \
    --steps=2000 \
    --batch_size=16 \
    --save_freq=500 \
    --log_freq=50 \
    --wandb.enable=true
```

```python
# @title 4. Ã‰valuer dans Meta-World (task assembly-v3, 3 Ã©pisodes) + enregistrer les vidÃ©os
!lerobot-eval \
    --policy.path=outputs/metaworld_smolvla/checkpoints/last/pretrained_model \
    --env.type=metaworld \
    --env.task=assembly-v3 \
    --eval.batch_size=1 \
    --eval.n_episodes=3 \
    --eval.recording=true \
    --eval.recording_repo_id=oreocookiedough/eval_metaworld
```

```python
# @title 5. PrÃ©parer les sorties pour le tÃ©lÃ©chargement
import os
import shutil

os.makedirs('/kaggle/working/output', exist_ok=True)
for sub in ['outputs', 'eval_logs']:
    src = f'/kaggle/working/lerobot/{sub}'
    if os.path.exists(src):
        shutil.move(src, f'/kaggle/working/output/{sub}')
!rm -rf /kaggle/working/lerobot
print('Outputs :', os.listdir('/kaggle/working/output'))
```
