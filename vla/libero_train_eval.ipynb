```python
# @title 3. Fine-tune SmolVLA sur LIBERO (commande officielle de la doc LeRobot)
!lerobot-train \
    --policy.type=smolvla \
    --policy.load_vlm_weights=true \
    --policy.push_to_hub=false \
    --dataset.repo_id=lerobot/libero \
    --dataset.video_backend=torchcodec \
    --output_dir=./outputs/libero_smolvla \
    --steps=2000 \
    --batch_size=16 \
    --save_freq=500 \
    --log_freq=50 \
    --wandb.enable=true
```
